```python
#este codigo é uma calculadora de operações basicas
#no terminal apenas insira os numeros e aperte enter
# -*- coding: utf-8 -*-


entrada1 = int (input('digite um valor: '))
entrada2 = int (input('digite um valor: '))
entrada3 = int (input('digite um valor: '))

print('De entrada na operação desejada de (1-4)')
print('1 soma')
print('2 subtração')
print('3 divisão')
print('4 multiplicação')

Opções= input('escolha uma operação de (1-4)')

if (Opções == '1'):
  print(entrada1, '+', entrada2, '=', entrada1+entrada2)
elif(Opções == '2'):
     print(entrada1, '-', entrada2, '=', entrada1-entrada2)
elif(Opções == '3'):
     print(entrada1, '/', entrada2, '=', entrada1/entrada2)
elif(Opções == '4'):
     print(entrada1, '*', entrada2, '=', entrada1*entrada2)
else:
      print('opção invalida')
```
